```python
import matplotlib.pyplot as plt
import numpy as np

# label, FP/min, x spread as a multiplicative factor, sensitivity, y spread, colour
points = [
    ("Model 1", 0.5,  1.6, 0.35, 0.06, "#1f77b4"),
    ("Model 2", 5.0,  1.5, 0.62, 0.07, "#d62728"),
    ("Model 3", 30.0, 1.4, 0.85, 0.05, "#2ca02c"),
]

fig, ax = plt.subplots(figsize=(6, 4.5))

t = np.linspace(0, 2 * np.pi, 200)
for label, fp, fp_fac, sens, sens_err, colour in points:
    ex = 10 ** (np.log10(fp) + np.log10(fp_fac) * np.cos(t))
    ey = sens + sens_err * np.sin(t)
    ax.fill(ex, ey, color=colour, alpha=0.18, linewidth=0, zorder=1)
    ax.plot(ex, ey, color=colour, alpha=0.7, linewidth=1.0, zorder=2)
    ax.plot(fp, sens, marker="o", markersize=7, linestyle="none",
            color=colour, zorder=3, label=label)

ax.set_xscale("log")
ax.set_xlim(1e-2, 1e2)
ax.set_xticks([1e-2, 1e-1, 1e0, 1e1, 1e2])
ax.set_ylim(0.0, 1.0)
ax.set_yticks(np.arange(0.0, 1.01, 0.2))
ax.set_xlabel("False positives per minute (FP/min)")
ax.set_ylabel("Sensitivity")
ax.grid(True, which="major", alpha=0.3, linewidth=0.5)
ax.tick_params(which="minor", length=0)
ax.legend(frameon=False, loc="lower right")

plt.show()
```
